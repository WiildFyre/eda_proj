```python
#Q1
import pandas as pd
import numpy as np
import math

# ---------------- Q1 ----------------
dataframe1 = pd.DataFrame({
    "ProductID": [101,102,103,104,105,106,107,108,109,110],
    "NameofProd": ["Laptop","Mouse","Keyboard","Monitor","Printer",
                   "Speaker","Webcam","SSD","RAM","Router"],
    "Price": [55000,700,1500,12000,9000,2500,1800,4500,3200,2200]
})

dataframe2 = pd.DataFrame({
    "ProductID": [101,102,103,104,105,106,107,108,109,110],
    "Company_Location": ["Delhi","Mumbai","Chennai","Bangalore",
                         "Hyderabad","Pune","Kolkata","Noida",
                         "Jaipur","Ahmedabad"]
})

dataframe3 = pd.DataFrame({
    "ProductID": [111,112,113,114,115,116,117,118,119,120],
    "NameofProd": ["Tablet","Scanner","Projector","Camera","Microphone",
                   "Headphones","UPS","PenDrive","HardDisk","Smartwatch"],
    "Price": [25000,8500,40000,30000,4500,3500,6000,900,6500,12000]
})

print(dataframe1, "\n")
print(dataframe2, "\n")
print(dataframe3, "\n")

# 1. Append dataframe1 and dataframe3
append_df = pd.concat([dataframe1, dataframe3], ignore_index=True)
print("\nAppended DataFrame:")
print(append_df)

# 2. Concatenate (column-wise)
concat_df = pd.concat([dataframe1, dataframe2[["Company_Location"]]], axis=1)
print("\nColumn-wise Concatenation:")
print(concat_df)

# 3. Merge dataframe1 and dataframe2
merged_df = pd.merge(dataframe1, dataframe2, on="ProductID")
print("\nMerged DataFrame:")
print(merged_df)

# 4. Join operations
inner_join = pd.merge(dataframe1, dataframe2, on="ProductID", how="inner")
print("\nInner Join:")
print(inner_join)

left_join = pd.merge(dataframe1, dataframe2, on="ProductID", how="left")
print("\nLeft Join:")
print(left_join)

right_join = pd.merge(dataframe1, dataframe2, on="ProductID", how="right")
print("\nRight Join:")
print(right_join)

full_join = pd.merge(dataframe1, dataframe2, on="ProductID", how="outer")
print("\nFull Outer Join:")
print(full_join)

# ---------------- Q2 ----------------
Arnav = np.array([44, 25, 40, 65, 80, 95, 110, 125, 140, 155], dtype=float)
print("\nOriginal Vector:")
print(Arnav)

def normalize(x, new_min, new_max):
    return ((x - x.min()) / (x.max() - x.min())) * (new_max - new_min) + new_min

norm_3 = normalize(Arnav, -3, 3)
print("\nNormalization [-3,+3]:")
print(norm_3)

norm_1 = normalize(Arnav, -1, 1)
print("\nNormalization [-1,+1]:")
print(norm_1)

minmax = normalize(Arnav, 0, 1)
print("\nMin-Max Normalization [0,1]:")
print(minmax)

# Z-score (sample std, same as R's scale())
zscore = (Arnav - Arnav.mean()) / Arnav.std(ddof=1)
print("\nZ-Score Transformation:")
print(zscore)

# Decimal scaling
j = math.ceil(math.log10(np.max(np.abs(Arnav)) + 1))
decimal_scale = Arnav / (10 ** j)
print("\nDecimal Scaling Normalization:")
print(decimal_scale)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from collections import Counter

sns.set_theme(style="darkgrid")

# Q1. Import dataset
data = pd.read_csv("edata.csv")

# Q2. First and last six records
print("First six records:")
print(data.head(6))
print("\nLast six records:")
print(data.tail(6))

# Q3. Total missing values
total_missing = data.isna().sum().sum()
print("\nTotal missing values:", total_missing)

# Q4. Records with at least one missing value
missing_records = data[data.isna().any(axis=1)]
print("\nFirst 10 records containing missing values:")
print(missing_records.head(10))

# Q5. Remove columns with more than 70% missing values
missing_percentage = data.isna().mean() * 100
print("\nPercentage of missing values in each column:")
print(missing_percentage)
data = data.loc[:, missing_percentage <= 70]
print("\nFirst 10 rows after removing columns:")
print(data.head(10))

# Q6. Replace missing numerical values with the mean
data_mean = data.copy()
num_cols = data_mean.select_dtypes(include=np.number).columns
data_mean[num_cols] = data_mean[num_cols].fillna(data_mean[num_cols].mean())
print("\nMean imputation:")
print(data_mean.head(10))

# Q7. Replace missing numerical values with the median
data_median = data.copy()
num_cols = data_median.select_dtypes(include=np.number).columns
data_median[num_cols] = data_median[num_cols].fillna(data_median[num_cols].median())
print("\nMedian imputation:")
print(data_median.head())

# Q8. Replace missing categorical values with the mode
data_mode = data.copy()
cat_cols = data_mode.select_dtypes(include="object").columns
for col in cat_cols:
    data_mode[col] = data_mode[col].fillna(data_mode[col].mode()[0])
print("\nMode imputation:")
print(data_mode.head(10))

# Q9. Identify duplicate rows
duplicate_rows = data[data.duplicated()]
print("\nDuplicate rows:")
print(duplicate_rows)

# Q10. Remove duplicate rows
data = data.drop_duplicates()
print("\nAfter removing duplicates:")
print(data.head())

# Q11. Identify columns with incorrect data types
print("\nStructure of the dataset:")
data.info()
print("\nData type of each column:")
print(data.dtypes)

# Q12. Convert character variables into factors (category)
data["Category"] = data["Category"].astype("category")
data["Sex"] = data["Sex"].astype("category")
print("\nStructure after conversion:")
data.info()

# Q13. Lowercase / uppercase
data["Category"] = data["Category"].astype(str).str.lower()
data["Sex"] = data["Sex"].astype(str).str.upper()
print("\nCategory values:", data["Category"].head(10).tolist())
print("Sex values:", data["Sex"].head(10).tolist())

# Q14. Boxplot to detect outliers
plt.figure(figsize=(6, 6))
sns.boxplot(y=data["ALP"])
plt.title("Boxplot for ALP")
plt.ylabel("ALP")
plt.show()

# Q15. Outliers using IQR method
Q1 = data["ALP"].quantile(0.25)
Q3 = data["ALP"].quantile(0.75)
IQR_value = Q3 - Q1
lower_limit = Q1 - 1.5 * IQR_value
upper_limit = Q3 + 1.5 * IQR_value
print("\nQ1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR_value)
print("Lower limit:", lower_limit)
print("Upper limit:", upper_limit)
outliers = data.loc[(data["ALP"] < lower_limit) | (data["ALP"] > upper_limit), "ALP"]
print("Outliers:", outliers.tolist())

# Q16. Remove extreme outliers (keeping NaN rows, as in the R code)
data_no_outliers = data[data["ALP"].isna() |
                        ((data["ALP"] >= lower_limit) & (data["ALP"] <= upper_limit))]
print("\nAfter removing outliers:")
print(data_no_outliers.head(10))

# Q17. Histogram
plt.figure(figsize=(8, 5))
plt.hist(data["Age"].dropna(), bins=range(15, 85, 5), color="steelblue", edgecolor="black")
plt.title("Histogram of Age")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
print("Interpretation: The histogram shows the distribution of Age across the observations.")

# Q18. Histogram with density curve
plt.figure(figsize=(8, 5))
sns.histplot(data["Age"].dropna(), bins=range(15, 85, 5), stat="density",
             color="lightblue", edgecolor="black")
sns.kdeplot(data["Age"].dropna(), color="black", linewidth=2)
plt.title("Histogram with Density Curve for Age")
plt.xlabel("Age")
plt.ylabel("Density")
plt.show()

# Q19. Violin plot
plt.figure(figsize=(7, 5))
sns.violinplot(x="Sex", y="Age", data=data, order=["F", "M"],
               color="white", linecolor="black", fill=False)
plt.title("Violin Plot of Age by Sex")
plt.xlabel("Sex")
plt.ylabel("Age")
plt.show()

# Q20. Dot plot for a small numerical dataset
small_data = data.iloc[:20]
counts = Counter(small_data["Age"].dropna())
xs, ys = [], []
for age, c in counts.items():
    for i in range(1, c + 1):
        xs.append(age)
        ys.append(i)
plt.figure(figsize=(7, 5))
plt.scatter(xs, ys, s=200, color="black")
plt.title("Dot Plot of Age")
plt.xlabel("Age")
plt.ylabel("Count")
plt.show()

# Q21. Frequency table
frequency_table = data["Category"].value_counts()
print("\nFrequency table:")
print(frequency_table)

# Q22. Proportion and percentage
proportion = data["Category"].value_counts(normalize=True)
print("\nProportion:")
print(proportion)
percentage = proportion * 100
print("\nPercentage:")
print(percentage)

# Q23. Bar chart
plt.figure(figsize=(9, 5))
data["Category"].value_counts().sort_index().plot(kind="bar", color="gray", edgecolor="black")
plt.title("Frequency of Each Category")
plt.xlabel("Category")
plt.ylabel("Frequency")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()

# Q24. Horizontal bar chart
plt.figure(figsize=(9, 5))
data["Category"].value_counts().sort_index().plot(kind="barh", color="gray", edgecolor="black")
plt.title("Horizontal Bar Chart of Categories")
plt.xlabel("Frequency")
plt.ylabel("Category")
plt.tight_layout()
plt.show()

# Q25. Percentage bar chart
plt.figure(figsize=(9, 5))
pct = data["Category"].value_counts(normalize=True).sort_index() * 100
pct.plot(kind="bar", color="gray", edgecolor="black")
plt.title("Percentage Bar Chart")
plt.xlabel("Category")
plt.ylabel("Percentage")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()

# Q26. Pie chart
pie_data = data["Sex"].value_counts()
plt.figure(figsize=(6, 6))
plt.pie(pie_data, labels=pie_data.index, autopct="%1.1f%%", startangle=90)
plt.title("Pie Chart of Sex")
plt.show()

# Q27. Lollipop chart
lollipop_data = data["Category"].value_counts().sort_values()
plt.figure(figsize=(9, 5))
plt.hlines(y=lollipop_data.index, xmin=0, xmax=lollipop_data.values, color="black")
plt.plot(lollipop_data.values, lollipop_data.index, "o", markersize=9, color="black")
plt.title("Lollipop Chart of Category Frequencies")
plt.xlabel("Frequency")
plt.ylabel("Category")
plt.tight_layout()
plt.show()

# Q28. Dot chart
dot_data = data["Category"].value_counts().sort_values()
plt.figure(figsize=(9, 5))
plt.plot(dot_data.values, dot_data.index, "o", markersize=7, color="black")
plt.title("Dot Chart of Category Frequencies")
plt.xlabel("Frequency")
plt.ylabel("Category")
plt.grid(axis="y")
plt.tight_layout()
plt.show()

# Q29. Density plot
plt.figure(figsize=(8, 5))
sns.kdeplot(data["Age"].dropna(), color="black")
plt.title("Density Plot of Age")
plt.xlabel("Age")
plt.ylabel("Density")
plt.show()

# Q30. Facet (one histogram per Sex group)
g = sns.FacetGrid(data, col="Sex", height=5, aspect=1.1, col_order=["F", "M"])
g.map(plt.hist, "Age", bins=range(15, 85, 5), color="gray", edgecolor="black")
g.set_axis_labels("Age", "Frequency")
g.figure.suptitle("Distribution of Age Across Sex Groups", y=1.03)
plt.show()

print("Interpretation:")
print("Male: The distribution shows the frequency of different age groups among male observations.")
print("Female: The distribution shows the frequency of different age groups among female observations.")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

sns.set_theme(style="whitegrid")

# 1. Load dataset
print("Answer 1")
data = pd.read_csv("edata.csv")
print(data.head(3))
data.info()
num = ["Age","ALB","ALP","ALT","AST","BIL","CHE","CHOL","CREA","GGT","PROT"]

# 2. Summary statistics (median imputation first, as in the R code)
print("\nAnswer 2")
for col in num:
    data[col] = data[col].fillna(data[col].median())

def e1071_skewness(x):
    # e1071 default type = 3
    n = len(x)
    m2 = np.mean((x - x.mean()) ** 2)
    m3 = np.mean((x - x.mean()) ** 3)
    g1 = m3 / m2 ** 1.5
    return g1 * ((n - 1) / n) ** 1.5

def e1071_kurtosis(x):
    # e1071 default type = 3 (excess kurtosis)
    n = len(x)
    m2 = np.mean((x - x.mean()) ** 2)
    m4 = np.mean((x - x.mean()) ** 4)
    g2 = m4 / m2 ** 2 - 3
    return (g2 + 3) * (1 - 1 / n) ** 2 - 3

summary_stats = pd.DataFrame({
    "Variable": num,
    "Mean": [data[c].mean() for c in num],
    "Median": [data[c].median() for c in num],
    "Minimum": [data[c].min() for c in num],
    "Maximum": [data[c].max() for c in num],
    "Range": [data[c].max() - data[c].min() for c in num],
    "Variance": [data[c].var() for c in num],
    "Standard_Deviation": [data[c].std() for c in num],
    "IQR": [data[c].quantile(0.75) - data[c].quantile(0.25) for c in num],
    "Skewness": [e1071_skewness(data[c].values) for c in num],
    "Kurtosis": [e1071_kurtosis(data[c].values) for c in num],
}, index=num)
pd.set_option("display.width", 250)
pd.set_option("display.max_columns", None)
print(summary_stats)

# 3. Store in structured data frame
print("\nAnswer 3")
print(summary_stats)

# 4. Frequency, relative frequency, Pie chart, Bar plot
print("\nAnswer 4")
category_freq = data["Category"].value_counts().sort_index()
category_relative_freq = category_freq / category_freq.sum()
print(category_freq)
print(category_relative_freq)

plt.figure(figsize=(7, 7))
plt.pie(category_freq, labels=category_freq.index, colors=sns.color_palette("hls", len(category_freq)))
plt.title("Q4: Pie Chart of Category Distribution")
plt.show()

plt.figure(figsize=(9, 5))
plt.bar(category_freq.index, category_freq.values, color="gray", edgecolor="black")
plt.title("Q4: Bar Plot of Category Distribution")
plt.xlabel("Category")
plt.ylabel("Frequency")
plt.xticks(rotation=30, ha="right")
plt.tight_layout()
plt.show()

# 5. Frequency table with bins, Histogram, Box plot
print("\nAnswer 5")
age_breaks = np.arange(np.floor(data["Age"].min()), np.ceil(data["Age"].max()) + 10, 10)
age_bins = pd.cut(data["Age"], bins=age_breaks, include_lowest=True, right=False)
age_frequency = age_bins.value_counts().sort_index()
print(age_frequency)

plt.figure(figsize=(8, 5))
plt.hist(data["Age"], bins=15, color="skyblue", edgecolor="white")
plt.title("Q5: Histogram of Age")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()

plt.figure(figsize=(5, 6))
plt.boxplot(data["Age"], patch_artist=True, boxprops=dict(facecolor="lightgreen"))
plt.title("Q5: Box Plot of Age")
plt.ylabel("Age")
plt.show()

# 6. Contingency table
print("\nAnswer 6")
contingency_table = pd.crosstab(data["Category"], data["Sex"])
print(contingency_table)
print("\nRow Percentages:")
print((contingency_table.div(contingency_table.sum(axis=1), axis=0) * 100).round(2))
print("\nColumn Percentages:")
print((contingency_table.div(contingency_table.sum(axis=0), axis=1) * 100).round(2))
print("\nOverall Percentages:")
print((contingency_table / contingency_table.values.sum() * 100).round(2))

# 7. Stacked bar plot
print("\nAnswer 7")
contingency_table.plot(kind="bar", stacked=True, figsize=(9, 5))
plt.title("7: Stacked Bar Plot: Category vs Sex")
plt.xlabel("Category")
plt.ylabel("Frequency")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()

# 8. Aggregation
print("\nAnswer 8")
age_by_category = data.groupby("Category")["Age"].agg(Mean="mean", Median="median", SD="std")
print("\nAge Statistics Grouped by Category:")
print(age_by_category)
print("\nMean Age:")
print(data.groupby("Category")["Age"].mean())
print("\nMedian Age:")
print(data.groupby("Category")["Age"].median())
print("\nStandard Deviation of Age:")
print(data.groupby("Category")["Age"].std())

# 9. Side-by-side box plot and violin plot
print("\nAnswer 9")
plt.figure(figsize=(10, 6))
sns.boxplot(x="Category", y="Age", data=data, hue="Category", legend=False)
plt.title("9: Side-by-Side Box Plot of Age by Category")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()

plt.figure(figsize=(10, 6))
sns.violinplot(x="Category", y="Age", data=data, hue="Category", legend=False, cut=0)
plt.title("9: Violin Plot of Age by Category")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()

# 10. Pearson and Spearman correlation, scatter plot with regression line
print("\nAnswer 10")
pearson_corr = data["ALB"].corr(data["ALT"], method="pearson")
spearman_corr = data["ALB"].corr(data["ALT"], method="spearman")
print("Pearson Correlation (ALB vs ALT):", pearson_corr)
print("Spearman Correlation (ALB vs ALT):", spearman_corr)

plt.figure(figsize=(8, 6))
sns.regplot(x="ALB", y="ALT", data=data,
            scatter_kws={"color": "steelblue", "alpha": 0.6},
            line_kws={"color": "red"})
plt.title("10: ALB vs ALT with Linear Regression Line")
plt.show()

# 11. Group summary and plots
print("\nAnswer 11")
group_summary_final = (data.groupby(["Category", "Sex"])["Age"]
                       .agg(Mean="mean", SD="std").reset_index())
print("\nClean Group Summary:")
print(group_summary_final)

g = sns.catplot(data=data, x="Sex", y="Age", hue="Sex", col="Category",
                col_wrap=3, kind="box", height=3.5, legend=False)
g.figure.suptitle("11: Faceted Box Plot of Age by Category and Sex", y=1.03)
plt.show()

pivot_mean = group_summary_final.pivot(index="Category", columns="Sex", values="Mean")
pivot_sd = group_summary_final.pivot(index="Category", columns="Sex", values="SD").fillna(0)
pivot_mean.plot(kind="bar", yerr=pivot_sd, capsize=4, figsize=(10, 6))
plt.title("11: Mean Age by Category and Sex")
plt.xlabel("Category")
plt.ylabel("Mean Age ± SD")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()

# 12. Scatter plot colored/shaped by category
print("\nAnswer 12")
plt.figure(figsize=(9, 6))
sns.scatterplot(data=data, x="ALB", y="ALT", hue="Category", style="Category", s=50, alpha=0.7)
plt.title("12: ALB vs ALT by Category")
plt.legend(bbox_to_anchor=(1.02, 1), loc="upper left")
plt.tight_layout()
plt.show()

print("Correlation (ALB, ALT) within each category:")
for cat, grp in data.groupby("Category"):
    print(cat, "->", round(grp["ALB"].corr(grp["ALT"]), 4))

# 13. Correlation matrix heatmap
print("\nAnswer 13")
cor_vars = ["Age", "ALB", "ALT", "AST"]
cor_matrix = data[cor_vars].corr(method="pearson")
print(cor_matrix)
plt.figure(figsize=(6, 5))
sns.heatmap(cor_matrix, annot=True, fmt=".2f", cmap="bwr", vmin=-1, vmax=1, center=0)
plt.title("13: Correlation Matrix Heatmap")
plt.show()

# 14. Pairs plot
print("\nAnswer 14")
pd.plotting.scatter_matrix(data[cor_vars], figsize=(9, 9), diagonal="hist",
                           c="steelblue", s=15)
plt.suptitle("14: Pairs Plot / Scatterplot Matrix")
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from scipy.cluster.hierarchy import linkage, dendrogram, fcluster
from scipy.spatial.distance import pdist

# ================= K-MEANS CLUSTERING =================

# 1. Load dataset
data = pd.read_csv("edata.csv")
print("Answer No. 1 - Dataset Loaded Successfully")

# 2. View dataset
print("\nAnswer No. 2 - Dataset")
print(data.head(3))
print("\nDataset Structure:")
data.info()
print("\nDataset Summary:")
print(data.describe(include="all"))

# 3. Missing values
print("\nAnswer No. 3 - Missing Values")
print(data.isna().sum())
print("\nTotal Missing Values:", data.isna().sum().sum())

# 4. Dependent variable distribution
print("\nAnswer No. 4 - Dependent Variable Distribution")
print(data["ALP"].value_counts().sort_index().head(10))

# 5. Numerical features only
print("\nAnswer No. 5 - Numerical Features Selected")
numeric_data = data.select_dtypes(include=np.number)
print(numeric_data.head(3))
print("\nNumerical Feature Names:")
print(list(numeric_data.columns))
numeric_data = numeric_data.dropna()
print("\nRows remaining after removing missing values:", len(numeric_data))

# 6. Standardize (sample std, same as R's scale())
print("\nAnswer No. 6 - Standardized Data")
scaled_data = (numeric_data - numeric_data.mean()) / numeric_data.std(ddof=1)
print(scaled_data.head(6))

# 7. K-Means with K = 2
print("\nAnswer No. 7 - K-Means Clustering with K = 2")
kmeans_2 = KMeans(n_clusters=2, n_init=25, random_state=123).fit(scaled_data)
print("Inertia (tot.withinss):", kmeans_2.inertia_)
print("Iterations:", kmeans_2.n_iter_)

# 8. K-Means results
print("\nAnswer No. 8 - K-Means Results")
print("Cluster Assignments (first 5):")
print(pd.Series(kmeans_2.labels_ + 1, index=np.arange(1, len(scaled_data) + 1)).head(5))
print("\nNumber of Samples in Each Cluster:")
print(pd.Series(kmeans_2.labels_ + 1).value_counts().sort_index())

# 9. Cluster centers
print("\nAnswer No. 9 - Cluster Centers")
print(pd.DataFrame(kmeans_2.cluster_centers_, columns=scaled_data.columns,
                   index=[1, 2]))

# 10. WCSS for K = 1 to 10
print("\nAnswer No. 10 - WCSS for K = 1 to 10")
wcss = []
for k in range(1, 11):
    km = KMeans(n_clusters=k, n_init=25, random_state=123).fit(scaled_data)
    wcss.append(km.inertia_)
wcss_table = pd.DataFrame({"K": range(1, 11), "WCSS": wcss})
print(wcss_table)

# 11. Elbow plot
print("\nAnswer No. 11 - Elbow Plot")
plt.figure(figsize=(8, 5))
plt.plot(range(1, 11), wcss, "o-", color="blue")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("WCSS")
plt.title("11: Elbow Method")
plt.show()

# 12. Final K-Means (K = 3)
print("\nAnswer No. 12 - Final K-Means Clustering")
optimal_k = 3
kmeans_final = KMeans(n_clusters=optimal_k, n_init=25, random_state=123).fit(scaled_data)
print("\nSelected Number of Clusters:", optimal_k)
print("\nNumber of Samples in Each Cluster:")
print(pd.Series(kmeans_final.labels_ + 1).value_counts().sort_index())
print("\nFinal Cluster Centers:")
print(pd.DataFrame(kmeans_final.cluster_centers_, columns=scaled_data.columns,
                   index=[1, 2, 3]))

# ================= HIERARCHICAL CLUSTERING =================

# 1. Load dataset
data_hc = pd.read_csv("edata.csv")

# 2. View dataset
print("\n\nDataset Summary:")
print(data_hc.describe(include="all"))

# 3. Missing values
print("\nAnswer No. 3 - Missing Values")
print(data_hc.isna().sum())
print("\nTotal Missing Values:", data_hc.isna().sum().sum())

# 4. Numerical attributes
print("\nAnswer No. 4 - Numerical Attributes")
numeric_hc = data_hc.select_dtypes(include=np.number).dropna()
print("\nNumerical Attribute Names:")
print(list(numeric_hc.columns))

# 5. Standardize
print("\nAnswer No. 5 - Standardized Data")
scaled_hc = (numeric_hc - numeric_hc.mean()) / numeric_hc.std(ddof=1)
print(scaled_hc.head(5))

# 6. Euclidean distance
print("\nAnswer No. 6 - Euclidean Distance")
distance_matrix = pdist(scaled_hc.values, metric="euclidean")
print(distance_matrix[:20])

# 7. Hierarchical clustering with Ward's method
print("\nAnswer No. 7 - Hierarchical Clustering using Ward's Method")
hc = linkage(distance_matrix, method="ward")
print("Linkage matrix shape:", hc.shape)
print("Number of objects:", scaled_hc.shape[0])

# 8. Dendrogram
print("\nAnswer No. 8 - Dendrogram")
plt.figure(figsize=(12, 6))
dendrogram(hc, no_labels=True)
plt.title("8: Hierarchical Clustering Dendrogram")
plt.xlabel("Observations")
plt.ylabel("Height")
plt.show()

# 9. Cut into 3 clusters
print("\nAnswer No. 9 - Cluster Assignments for 3 Clusters")
hc_clusters = fcluster(hc, t=3, criterion="maxclust")
print(pd.Series(hc_clusters, index=np.arange(1, len(hc_clusters) + 1)).head(35))

# 10. Samples per cluster
print("\nAnswer No. 10 - Number of Samples in Each Cluster")
print(pd.Series(hc_clusters).value_counts().sort_index())
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA

# 1. Load dataset
data = pd.read_csv("edata.csv")

# 2. Rows and columns
print("Number of rows:", data.shape[0])
print("Number of columns:", data.shape[1])

# 3. Numeric columns, clean, normalize
data_num = data.select_dtypes(include=np.number).dropna()
data_scaled = (data_num - data_num.mean()) / data_num.std(ddof=1)
print(data_scaled.head())

# 4. PCA and summary
pca = PCA()
scores = pca.fit_transform(data_scaled)

sdev = np.sqrt(pca.explained_variance_)
prop_var = pca.explained_variance_ratio_
cum_var = np.cumsum(prop_var)
pc_names = [f"PC{i+1}" for i in range(len(sdev))]

print("\nImportance of components:")
summary = pd.DataFrame({
    "Standard deviation": sdev,
    "Proportion of Variance": prop_var,
    "Cumulative Proportion": cum_var,
}, index=pc_names).T
print(summary.round(5))

# 5. Scree plot with labels, y-axis 0 to 60
var_percent = prop_var * 100
plt.figure(figsize=(9, 5))
bars = plt.bar(pc_names, var_percent, color="gray", edgecolor="black")
for i, v in enumerate(var_percent):
    plt.text(i, v + 0.8, f"{v:.2f}", ha="center")
plt.ylim(0, 60)
plt.title("Scree Plot")
plt.xlabel("Principal Components")
plt.ylabel("Variance (%)")
plt.show()

print("\nVariance percentage of each component:")
print(np.round(var_percent, 2))

# 6. PCA biplot
plt.figure(figsize=(9, 8))
plt.scatter(scores[:, 0], scores[:, 1], s=8, color="lightgray")
loadings = pca.components_.T[:, :2]
scale_factor = np.abs(scores[:, :2]).max() * 0.8
for i, name in enumerate(data_scaled.columns):
    plt.arrow(0, 0, loadings[i, 0] * scale_factor, loadings[i, 1] * scale_factor,
              color="red", head_width=0.1, alpha=0.7)
    plt.text(loadings[i, 0] * scale_factor * 1.08,
             loadings[i, 1] * scale_factor * 1.08,
             name, color="red", fontsize=9, ha="center")
plt.axhline(0, color="black", linewidth=0.5)
plt.axvline(0, color="black", linewidth=0.5)
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.title("PCA Biplot")
plt.show()

# 7. PCA scores
pca_scores = pd.DataFrame(scores, columns=pc_names)
print("\nFirst six PCA scores:")
print(pca_scores.head(6))

# 8. Scatter plot of PC1 vs PC2
plt.figure(figsize=(8, 5))
plt.scatter(pca_scores["PC1"], pca_scores["PC2"], color="blue", marker="o", s=20)
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.title("PCA Plot - 23BDS0044")
plt.show()
```
